## 10.1 MultiIndex & Indexation Avancée

```python
# Créez un MultiIndex : Région → Département → Grade → Année d’embauche
multi_index = df.set_index(['region', 'department', 'grade', 'hire_year']).index
print(multi_index[4:15])
```

```python
# Calcul des statistiques salariales à chaque niveau du MultiIndex
salary_stats = df.groupby(['region', 'department', 'grade', 'hire_year'])['base_salary'].agg(['mean', 'sum', 'max', 'min', 'std'])
print(salary_stats[4:15])
```

```python
# Utilisez xs() pour extraire les Managers d’Europe embauchés après 2020

# Sélection des managers Europe
managers_europe = salary_stats.xs(('Europe', 'Manager'), level=['region', 'grade'])

# Filtrer les années d'embauche supérieures à 2020 sur l'index hire_year
managers_europe_2020 = managers_europe.loc[managers_europe.index.get_level_values('hire_year') > 2020]

print(managers_europe_2020)
```

### Sélection Multi-Niveaux

```python
idx = pd.IndexSlice

# Création du MultiIndex
df_multi = df.set_index(['region', 'department', 'grade', 'hire_year'])

# Sélection par département "Technologie" OU "Finance"
selected_dept = df_multi.loc[idx[:, ("Technologie", "Finance"), :, :], :]

print("Sélection par département (Technologie ou Finance) :")
print(selected_dept)
```

```python
# Sélection par grade "Senior" OU "Lead"
selected_grade = df_multi.loc[idx[:, :, ("Senior", "Lead"), :], :]

print("\nSélection par grade (Senior ou Lead) :")
print(selected_grade)
```

```python
# Filtrer les employés avec un score de performance supérieur à 7.5
selected_per = selected_grade[selected_grade['performance_score'] > 7.5]
print("\nEmployés avec un score de performance > 7.5 :")
print(selected_per["performance_score"])
```
