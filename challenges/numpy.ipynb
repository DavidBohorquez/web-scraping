## 1. Primitives importantes de Numpy

1.1 Statisques de base

```python
import numpy as np

data = np.array([1, 2, 3, 4])
mean_value = data.mean()
print("Moyenne :", mean_value)
```

```python
total = data.sum()
print("Somme :", total)
```

```python
variance = data.var()
print("Variance :", variance)
```

1.2 Manipulation des tableaux

```python
data = np.array([1, 2, 3, 4, 5, 6])
reshaped = data.reshape(2, 3)
print(reshaped)
```

```python
data = np.array([[1,2],[3,4]])
print(data)
flattened = data.flatten()
print("===================\n",flattened)
```

```python
transposed = data.transpose()
print(transposed)
```

```python
data1 = np.array([1,2])
data2 = np.array([3,4])
h_combined = np.hstack((data1, data2))
print("Horizontalement empilé :", h_combined)
v_combined = np.vstack((data1, data2))
print("Verticalement empilé :\n", v_combined)
```

```python
data = np.array([10, 20, 30, 40,])
last_element = data[-1]
print("Dernier élément :", last_element)
```

```python
data = np.array([10, 20, 30, 40, 50])
slice_data = data[::2]
print("Éléments aux indices pairs :", slice_data)
```

```python
data = np.array([1, 2, 3, 4])
copied_data = data.copy()
copied_data[0] = 99
print("Original :", data)
print("Copié :", copied_data)
```

1.3 Affichage des primier et dernier élements du dataset

```python
dataset = np.arange(1, 101)
print("Premiers éléments :", dataset[:10])
print("Derniers éléments :", dataset[-10:])
```

1.5 Indexation avancée

```python
# Fancy Indexing
data = np.array([10, 20, 30, 40, 50])
indices = [0, 2, 4]
selected = data[indices]
print(selected)
```

```python
# masking
data = np.array([1, 2, 3, 4, 5])
filtered = data[data > 3]
print(filtered)
```

```python
# where()
data = np.array([10, 20, 30, 40, 50])
condition = data > 30
resutl = np.where(condition)
print(resutl)
```

```python
import numpy as np
import pandas as pd
data = np.array([
    {"Nom": "Alice", "Classe": "Guerrier", "Niveau": 5},
    {"Nom": "Bob",  "Classe": "Mage", "Niveau": 3},
    {"Nom": "Charlie", "Classe": "Guerrier", "Niveau": 7},
])
data = np.array([
    ("Alice", "Guerrier", 5),
    ("Bob", "Mage", 3),
    ("Charlie", "Guerrier", 7)
], dtype=[('Nom','S'), ('Classe','S'), ('Niveau','i4')])

# Sélectionner tout les guerriers
guerriers = data[data["Classe"] == "Guerrier"]
print(guerriers)
```

```python
import numpy as np
import pandas as pd
data = np.array([
    {"Nom": "Alice", "Classe": "Guerrier", "Niveau": 5},
    {"Nom": "Bob",  "Classe": "Mage", "Niveau": 3},
    {"Nom": "Charlie", "Classe": "Guerrier", "Niveau": 7},
])
data = pd.DataFrame(data.tolist())
# Sélectionner tout les guerriers
guerriers = data[data["Classe"] == "Guerrier"]
print(guerriers)
```

```python
import numpy as np

# Création des tableaux
A = np.array([1, 2, 3])        # Shape: (3,)
B = np.array([[1], [2], [3]])  # Shape: (3, 1)

print("Tableau A:")
print(f"Shape: {A.shape}")
print(f"Valeurs: {A}")
print("\nTableau B:")
print(f"Shape: {B.shape}")
print(f"Valeurs:\n{B}")

# Opération avec broadcasting
result = A + B
print("\nRésultat A + B:")
print(f"Shape: {result.shape}")
print(f"Valeurs:\n{result}")
```
